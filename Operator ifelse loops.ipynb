# if-else in python

```python
email=input("Enter Email : ")
password=input("Enter Password : ")

if email == "gopal@123gmail.com" and password=='1234':
    print('welcome')

elif email == "gopal@123gmail.com" and password!='1234':
    print("Incorrect Password")
    password=input("Enter your password")
    if password =='1234':
        print('welcome! finally')
    else:
        print("tum se na hu payegaa")

else:
    print('not correct')
```

```python
# min of three number 
a=int(input("enter 1st num : "))
b=int(input("enter 1st num : "))
c=int(input("enter 1st num : "))

if a<b and a<c:
    print('smallest ',a)
elif b<c:
    print('smallest',b)
else:
    print('smallest',c)
```
