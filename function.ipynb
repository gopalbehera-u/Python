# Variable Length argument

```python
def sum(a,b):
    c=a+b
    print(c)

sum(34,55)
```

```python
def sum(a,b):
    c=a+b
    print(c)

sum(3,4,5,7)
```

```python
def sum(a,*b):# it returns tuple 
    c=a+b
    print(c)


sum(5,6,7,8)
```

```python
def sum(a,*b):
    print(type(a))
    print(type(b))

sum(45,678,998,7)
```

```python
def sum(a,*b):
    c=a
    for i in b:
        c=c+i
    print(c)

sum(2,4,4,5,5,)
```

```python
def myfun(*args,**kwargs):
    print("Non -Keyword Argument (*agrs) : ")
    for arg in args:
        print(arg)
    
    print("Keyword Argument (**kwargs) : ")
    for key,value in kwargs.items():
        print(f"{key} == {value}")


myfun("Hey","Welcome", "to my room",first="Geeks",mid="for",last="Geeks")
```
