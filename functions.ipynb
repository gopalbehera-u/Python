# *args and **kwargs

```python
# args
def multiply(*args):
    product=1
    for i in args:
        product=product*i
    return product
#args stores in tuples
```

```python
multiply(1,2,3,4,4)
```

```python
is_even.__doc__ # aceesing doc string in function
```

```python
#kwargs
def display(**kwargs):
    for key,value in kwargs.items():
        print(key,'->',value)
    print(kwargs)
```

```python
display(india='delhi',nepal='kathmandu',srilanka='colombo')
```

# Variable scope

```python
def g(y): # y=local var
    print(x)
    print(x+1)
x=5#global var
g(x)
print(x)
```

```python
def f(y):
    x=1# local var
    x+=1
    print(x)
x=5# global var
f(x)
print(x)
```

```python
def h(y):
    x+=1
x=5
h(x)
print(x)
```

```python
def h(y):
    global x
    x+=1
x=5
h(x)
print(x)
```

```python
def f(x):
    x=x+1
    print("in f(x): x = ",x)
    return x
x=3
z=f(x)
print("in main program scope : z= ",z)
print(" in main program scope : x = ",x)
```

# Higher Order Function

```python
def sqr(x):
    return x**2
# higher order function
def transform(f,l):
    output=[]
    for i in l:
        output.append(f(i))
    print(output)

l=[1,2,3,4,5,6]
transform(sqr,l)
```

```python
def sqr(x):
    return x**2
# higher order function
def transform(f,l):
    output=[]
    for i in l:
        output.append(f(i))
    print(output)

l=[1,2,3,4,5,6]
transform(lambda x:x**3,l)
```
