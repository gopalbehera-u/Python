# pattern 1 

```python
rows=int(input("Enter number of rows "))
for i in range(rows):
    for j in range(i):
        print("* ",end='')
    print()
```

```python
rows=int(input("enter num of rows : "))
for i in range(rows):
    for j in range(i):
        print(j,end=' ')
    print()
```

# break statement 

```python
for i in range(9):
    if i==7:
        break
    print(i)
```

```python
lower=int(input("Enter number : "))
higher=int(input("enter number  : "))

for i in range(lower,higher+1):
    for j in range(2,i):
        if i%j==0:
            break
    else:
        print(i)
```

# continue 

```python
for i in range(10):
    if i==7:
        continue
    print(i)
```
