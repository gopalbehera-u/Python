```python
class computer():
    def __init__(self): # it call the method autometically when object is created 
                        # __init__ constructor 
                        # it main purpose to initialize the object attribute (instance class)
        print('i5','8gb')
    

c1=computer()
# c1.__init__()
```

```python
class computer():
    def __init__(self,cpu,ram):
        self.ram=ram
        self.cpu=cpu
    
    def config(self):
        print('config is ',self.cpu,self.ram)

com1=computer('i4',4)
com2=computer('i7',5)
com3=computer('i9',8)


com1.config()
com2.config()
com3.config()
```

```python
class computer():
    def __init__(self,cpu,mem,ram):
        self.cpu=cpu
        self.mem=mem
        self.ram=ram
    def config(self):
        print("config is ",self.cpu,self.mem,self.ram)


com1=computer('i5',8,255)
com2=computer('i5',8,255)
com3=computer('i5',8,255)

com1.config()
com2.config()
com3.config()


print(id(com1))
print(id(com2))
print(id(com3))
```

```python
class computer():
    def __init__(self):
        self.name='akash'
        self.age=90
    def update(self):
        self.age=30
        self.name='ram'

c1 = computer()
c2 = computer()

c1.name='gopal'
c2.age=65

print(c1.name, c1.age)

c1.update()

print(c1.name, c1.age)

```
