# Encapsulation

```python
class car:
    def __init__(self):
        self.mil=10 # instance var
        self.com='bmw'

c1=car()
c2=car()

print(c1.com,c1.mil)
print(c2.mil,c2.com)
```

# instance method 

```python
class student:
    school='nit'
    def __init__(self,m1,m2,m3):
        self.m1=m1
        self.m2=m2
        self.m3=m3
    def avg(self): # instance method 
        # the method which use instance var ,perform some operation 
        # take self as its first parameter 
        return (self.m1+self.m2+self.m3)/3

s1=student(10,20,30)
s2=student(20,50,20)


print(s1.avg())
print(s2.avg())
```

# class method

```python
class student():
    school='iit'
    def __init__(self,m1,m2,m3):
        self.m1=m1
        self.m2=m2
        self.m3=m3
    def avg(self):
        return (self.m1+self.m2+self.m3)/3
    
    def info(cls):
        # class method alwys return class var 
        return cls.school

s1=student(10,20,30)
s2=student(5,15,10)

print(s1.avg())
s1.info()
```

```python
class student():
    school='iit'
    def __init__(self,m1,m2,m3):
        self.m1=m1
        self.m2=m2
        self.m3=m3
    def avg(self):
        return (self.m1+self.m2+self.m3)/3
    @classmethod
    def info(cls):
        # A class method is a method that works
        #  with class variables instead of instance variables.
        return cls.school

s1=student(10,20,30)
s2=student(5,15,10)

print(s1.avg())
s1.info()
```

# static method 

```python
class student():
    school='nit'

    def __init__(self,m1,m2,m3):
        self.m1=m1
        self.m2=m2
        self.m3=m3
    def avg(self):
        return (self.m1+self.m2+self.m3)/3
    
    @classmethod
    def infoschol(cls):
        return cls.school
    
    @staticmethod
    # static method neither works with class var nor instance method 
    def info():
        print('this is studet class')

s1=student(10,20,30)
s2=student(10,27,43)

print(s1.avg())
print(s2.avg())


print(s1.infoschol())
s1.info()
```

# inner class 

```python
class student():
    def __init__(self,name,roll):
        self.name=name
        self.roll=roll
    def show(self):
        print(self.name,self.roll)

s1=student('a',1)
s2=student('b',2)

print(s1.name,s1.roll)
print(s2.name,s2.name)

s1.show()
```

# multi inheritance 

```python
class A:
    def feature1(self):
        print('feature1 is hii')
    def feature2(self):
        print('feature2 is hii')
class B(A):
    def feature3(self):
       print("feature3 is hello")
    def feature4(self):
        print("feature4 is hello")

class C(B):
    def feature5(self):
        print("feature5 is how are u ")



c1=C()

print(c1.feature1())
print(c1.feature2())
print(c1.feature3())
print(c1.feature4())
print(c1.feature5())
```

```python
class A:
    def feature1(self):
        print('feature1 is hii')
    def feature2(self):
        print('feature2 is hii')
class B():
    def feature3(self):
       print("feature3 is hello")
    def feature4(self):
        print("feature4 is hello")

class C(A,B):
    def feature5(self):
        print("feature5 is how are u ")



c1=C()

print(c1.feature1())
print(c1.feature2())
print(c1.feature3())
print(c1.feature4())
print(c1.feature5())
```
