```python
 # Write a Python program to count the number of vowels in a string.

str='Python Programming'

v=['a', 'e', 'i', 'o', 'u']
count=0
for i in  str.lower()  : 
    if i in v :
        count+=1
print(count)
```

```python
# Write a Python program to find the largest number in a list.

nums=[12,45,3,67,34]
l=nums[0]
for i in nums:
    if i>l:
        l=i
print(l)
```

```python
nums = [12, 45, 3, 67, 34]
large=nums[0]
seclarge=float('-inf')
for num in nums:
    if num > large:
        large=num
    elif num > seclarge and num !=large:
        seclarge=large
print("Largest : ",large)
print("secod largest : ",seclarge)
```

```python
# Write a Python program to find the smallest number in a list
nums=[12,45,3,67,34]
smallnum=nums[0]
for num in nums:
    if smallnum > num:
        smallnum=num
print(smallnum)
```
